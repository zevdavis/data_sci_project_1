```python
import nltk
import string
```

```python
nltk.download("stopwords")
```

```python
stopwords = nltk.corpus.stopwords.words("english")
```

```python
def clean_reviews(text):
  text = text.lower()
  text = text.translate(str.maketrans("", "", string.punctuation))
  cleaned = []
  for word in text.split():
    if word not in stopwords:
      cleaned.append(word)
  return cleaned
```

```python
clean_reviews("My name is Zev Martin Davis I am 21 years old I like to solve math problems and make music, and my favorite color is red.")
```

```python
dissatisfied_dj_ratings["content"].apply(clean_reviews)
```
