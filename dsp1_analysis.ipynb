```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/spotify_reviews_cleaned")
```

```python
dj_df = df[df["content"].str.contains("dj", case=False)]
dj_df.tail()
```

```python
dj_df = dj_df[dj_df["at"] >= "2023-02-22"]
dj_df.head(1000)
```

```python
dj_rating_count = dj_df["score"].value_counts().sort_index()
dj_rating_count.head()
```

```python
import matplotlib.pyplot as plt
```

```python
score = ["1", "2", "3", "4", "5"]
count = [137, 78, 90, 107, 155]
plt.figure(figsize=(8, 5))
bars = plt.bar(score, count, width=0.8)
for bar in bars:
  height = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2.0, height + .1, f'{height}', ha='center', va='bottom', fontsize=10)
plt.title("Spotify Rating Count Based on DJ X")
plt.xlabel("Rating (Out of 5)")
plt.ylabel("Rating Count")
plt.show()
```
